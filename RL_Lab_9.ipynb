```python
import numpy as np

def run_assignment_1():
    print("=" * 60)
    print("      ASSIGNMENT 1: ALGEBRAIC BELLMAN EXPECTATION SOLVER     ")
    print("=" * 60)

    # -----------------------------------------------------------------
    # Part A & B: Setup and Algebraic Solution of Linear Equations
    # -----------------------------------------------------------------
    # System of Equations derived from manual calculations:
    # 1. V(S0) - 0.81*V(S1) = -1.0
    # 2. -0.09*V(S0) + V(S1) = 89.9

    # Coefficient matrix A and constant vector B
    A = np.array([
        [1.0, -0.81],
        [-0.09, 1.0]
    ])
    B = np.array([-1.0, 89.9])

    # Directly solve simultaneous linear equations
    state_values = np.linalg.solve(A, B)
    V_S0, V_S1 = state_values[0], state_values[1]

    # -----------------------------------------------------------------
    # Part C: Action Value Computations using State Values
    # -----------------------------------------------------------------
    gamma = 0.9

    Q_S0_Down = -1 + gamma * V_S1
    Q_S0_Right = -1 + gamma * 0  # Wall is terminal with future value 0
    Q_S1_Right = 100 + gamma * 0  # Goal is terminal with future value 0
    Q_S1_Up = -1 + gamma * V_S0

    # -----------------------------------------------------------------
    # Verification Step: V(s) = sum_a pi(a|s) * Q(s,a)
    # -----------------------------------------------------------------
    V_S0_verified = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
    V_S1_verified = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

    # -----------------------------------------------------------------
    # Output and Summary Display
    # -----------------------------------------------------------------
    print("\n[PART B & C] RESULTS SUMMARY TABLE")
    print("-" * 45)
    print(f"{'State / Action Variable':<25} | {'Calculated Value':<15}")
    print("-" * 45)
    print(f"{'Vπ(S0)':<25} | {V_S0:.2f}")
    print(f"{'Vπ(S1)':<25} | {V_S1:.2f}")
    print(f"{'Qπ(S0, Down)':<25} | {Q_S0_Down:.2f}")
    print(f"{'Qπ(S0, Right)':<25} | {Q_S0_Right:.2f}")
    print(f"{'Qπ(S1, Right)':<25} | {Q_S1_Right:.2f}")
    print(f"{'Qπ(S1, Up)':<25} | {Q_S1_Up:.2f}")
    print("-" * 45)

    print("\n[PART D] VERIFICATION CHECKS")
    print(f"Calculated Vπ(S0): {V_S0:.4f} vs Verified via Policy Weighting: {V_S0_verified:.4f}")
    print(f"Calculated Vπ(S1): {V_S1:.4f} vs Verified via Policy Weighting: {V_S1_verified:.4f}")



if __name__ == "__main__":
    run_assignment_1()
```

```python
import numpy as np

def run_assignment_2():
    print("=" * 60)
    print("      ASSIGNMENT 2: ITERATIVE POLICY EVALUATION SOLVER      ")
    print("=" * 60)

    # -----------------------------------------------------------------
    # Part B: Iterative Policy Evaluation Setup
    # -----------------------------------------------------------------
    gamma = 0.9
    theta = 0.000001

    # State value initializations
    V_S0 = 0.0
    V_S1 = 0.0

    iteration = 0
    delta = 1.0

    # Iteration loop
    while delta >= theta:
        iteration += 1
        V_old_S0 = V_S0
        V_old_S1 = V_S1

        # Bellman expectation equation updates based on stochastic model
        # Wall and Goal states have future value 0
        V_S0_new = 0.1 * (-1 + gamma * 0) + 0.9 * (-1 + gamma * V_old_S1)
        V_S1_new = 0.9 * (100 + gamma * 0) + 0.1 * (-1 + gamma * V_old_S0)

        # Determine maximum change absolute value
        delta = max(abs(V_S0_new - V_old_S0), abs(V_S1_new - V_old_S1))

        # Commit state values update
        V_S0 = V_S0_new
        V_S1 = V_S1_new

    # -----------------------------------------------------------------
    # Part C: Action Values calculation after convergence
    # -----------------------------------------------------------------
    # Alternative actions transition definitions:
    # S0 Right -> Wall (1.0) | S1 Up -> S0 (1.0)
    Q_S0_Down = 0.1 * (-1 + gamma * 0) + 0.9 * (-1 + gamma * V_S1)
    Q_S0_Right = 1.0 * (-1 + gamma * 0)
    Q_S1_Right = 0.9 * (100 + gamma * 0) + 0.1 * (-1 + gamma * V_S0)
    Q_S1_Up = 1.0 * (-1 + gamma * V_S0)

    # -----------------------------------------------------------------
    # Output and Final Results Table
    # -----------------------------------------------------------------
    print(f"\nConvergence Reached At Iteration Count: {iteration}")
    print("\n[PART E] RESULTS SUMMARY TABLE")
    print("-" * 45)
    print(f"{'State / Action Variable':<25} | {'Calculated Value':<15}")
    print("-" * 45)
    print(f"{'Vπ(S0)':<25} | {V_S0:.2f}")
    print(f"{'Vπ(S1)':<25} | {V_S1:.2f}")
    print(f"{'Qπ(S0, Down)':<25} | {Q_S0_Down:.2f}")
    print(f"{'Qπ(S0, Right)':<25} | {Q_S0_Right:.2f}")
    print(f"{'Qπ(S1, Right)':<25} | {Q_S1_Right:.2f}")
    print(f"{'Qπ(S1, Up)':<25} | {Q_S1_Up:.2f}")
    print("-" * 45)

    print("\n[PART C.5] POLICY-CONSISTENCY CHECKS")
    print(f"Vπ(S0) matches Qπ(S0, Down)  : {np.isclose(V_S0, Q_S0_Down)}")
    print(f"Vπ(S1) matches Qπ(S1, Right) : {np.isclose(V_S1, Q_S1_Right)}")


if __name__ == "__main__":
    run_assignment_2()
```
