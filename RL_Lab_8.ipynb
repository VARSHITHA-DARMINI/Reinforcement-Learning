```python
# ==========================================================
# ASSIGNMENT 8
# Calculate State Value and Action Value
# 2 x 3 Grid World
# ==========================================================

print("==============================================")
print("     REINFORCEMENT LEARNING - ASSIGNMENT 8")
print("==============================================")

# ----------------------------------------------------------
# Grid World
# ----------------------------------------------------------
#
#       S0       S1        G
#       -------------------
#       X        S3        S4
#
# S0 = (0,0)
# S1 = (0,1)
# G  = (0,2)
# X  = (1,0)
# S3 = (1,1)
# S4 = (1,2)
#
# ----------------------------------------------------------

# Environment specification
gamma = 0.9
normal_reward = -1
goal_reward = 10
obstacle_reward = -5

# State positions
positions = {
    "S0": (0, 0),
    "S1": (0, 1),
    "G":  (0, 2),
    "X":  (1, 0),
    "S3": (1, 1),
    "S4": (1, 2)
}

# Policy given in the assignment
policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

# ----------------------------------------------------------
# Transition function
# ----------------------------------------------------------

def transition(state, action):

    row, col = positions[state]

    if action == "Up":
        new_row, new_col = row - 1, col

    elif action == "Down":
        new_row, new_col = row + 1, col

    elif action == "Left":
        new_row, new_col = row, col - 1

    elif action == "Right":
        new_row, new_col = row, col + 1

    # If outside grid, agent remains in same state
    if new_row < 0 or new_row > 1 or new_col < 0 or new_col > 2:
        return state

    # Obstacle cannot be entered
    if (new_row, new_col) == positions["X"]:
        return state

    # Find state at new position
    for s, pos in positions.items():
        if pos == (new_row, new_col):
            return s

    return state


# ----------------------------------------------------------
# Reward function
# ----------------------------------------------------------

def get_reward(next_state):

    if next_state == "G":
        return goal_reward

    elif next_state == "X":
        return obstacle_reward

    else:
        return normal_reward


# ----------------------------------------------------------
# Calculate State Value WITHOUT Bellman Equation
# ----------------------------------------------------------

def state_value(start_state):

    state = start_state
    total_return = 0
    discount = 1

    path = []
    rewards = []

    # Follow the given policy until goal
    while state != "G":

        path.append(state)

        action = policy[state]

        next_state = transition(state, action)

        reward = get_reward(next_state)

        rewards.append(reward)

        total_return += discount * reward

        discount *= gamma

        state = next_state

    path.append("G")

    return total_return, path, rewards


# ----------------------------------------------------------
# State Value for S0
# ----------------------------------------------------------

value_S0, path, rewards = state_value("S0")

print("\nGrid World:")
print("+------+------ +------+")
print("|  S0  |  S1  |   G  |")
print("+------+-------+------+")
print("|   X  |  S3  |  S4  |")
print("+------+-------+------+")

print("\nPolicy:")
for state, action in policy.items():
    print(state, ":", action)

print("\nPath followed from S0:")
print(" -> ".join(path))

print("\nRewards:")
print(rewards)

print("\nDiscount Factor (gamma):", gamma)

print("\nState Value of S0:")
print("V(S0) =", value_S0)


# ----------------------------------------------------------
# Action Value Q(S0, Up)
# ----------------------------------------------------------

def action_value(start_state, first_action):

    # First take the specified action
    next_state = transition(start_state, first_action)

    first_reward = get_reward(next_state)

    # Discounted first reward
    total_return = first_reward

    discount = gamma

    # Continue following policy
    state = next_state

    # Prevent infinite loop
    max_steps = 100
    steps = 0

    while state != "G" and steps < max_steps:

        action = policy[state]

        next_state = transition(state, action)

        reward = get_reward(next_state)

        total_return += discount * reward

        discount *= gamma

        state = next_state

        steps += 1

    return total_return


q_S0_up = action_value("S0", "Up")

print("\n==============================================")
print("Action Value for S0 and action UP")
print("==============================================")

print("Q(S0, Up) =", q_S0_up)
```
