```python
import pandas as pd

# Assignment 11: Policy Iteration in a Deterministic Grid World
# Course Code: 23CS201SE403
# Course: Reinforcement Learning

# 1. Define the MDP

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 0.001

normal_reward = -1
wall_reward = -1
goal_reward = 100

terminal_state = 'G'

print("1. MDP INFORMATION")
print("------------------")
print("States:", states)
print("Actions:", actions)
print("Normal Movement Reward:", normal_reward)
print("Wall Reward:", wall_reward)
print("Goal Reward:", goal_reward)
print("Discount Factor:", gamma)
print("Terminal State:", terminal_state)

print("\nGRID WORLD")
print("S0   | Wall")
print("S1   | G")


# 2. Define transitions
# Grid:
# S0   | Wall
# S1   | G

transitions = {
    'S0': {
        'Up': ('S0', wall_reward),
        'Down': ('S1', normal_reward),
        'Left': ('S0', wall_reward),
        'Right': ('S0', wall_reward)
    },
    'S1': {
        'Up': ('S0', normal_reward),
        'Down': ('S1', wall_reward),
        'Left': ('S1', wall_reward),
        'Right': ('G', goal_reward)
    }
}


# 3. Display and verify transitions

print("\n2. TRANSITION MAPPING")
print("---------------------")

for state in ['S0', 'S1']:
    for action in actions:
        next_state, reward = transitions[state][action]

        print(
            f"State: {state}, Action: {action}, "
            f"Next State: {next_state}, Reward: {reward}"
        )

        assert next_state in states

print("All transitions are defined.")


# 4. Initialize policy and state values

policy = {
    'S0': 'Up',
    'S1': 'Up'
}

V = {
    'S0': 0.0,
    'S1': 0.0,
    'G': 0.0
}

print("\n3. INITIAL POLICY")
print("-----------------")
print(policy)

print("\nINITIAL STATE VALUES")
print(V)


# 5. Policy Iteration

iteration = 0
policy_stable = False

while not policy_stable:

    iteration += 1

    print(f"\n========== POLICY ITERATION {iteration} ==========")

    # POLICY EVALUATION

    print("\nPolicy Evaluation")

    while True:
        old_V = V.copy()
        new_V = old_V.copy()

        for state in ['S0', 'S1']:
            action = policy[state]
            next_state, reward = transitions[state][action]

            new_V[state] = reward + gamma * old_V[next_state]

        new_V['G'] = 0.0

        delta = max(
            abs(new_V[state] - old_V[state])
            for state in states
        )

        V = new_V

        print(
            f"V(S0)={V['S0']:.4f}, "
            f"V(S1)={V['S1']:.4f}, "
            f"V(G)={V['G']:.4f}, "
            f"Delta={delta:.4f}"
        )

        if delta < theta:
            break

    # ACTION-VALUE CALCULATION

    Q = {}

    for state in ['S0', 'S1']:
        Q[state] = {}

        for action in actions:
            next_state, reward = transitions[state][action]

            Q[state][action] = reward + gamma * V[next_state]

    # POLICY IMPROVEMENT

    print("\nPolicy Improvement")

    old_policy = policy.copy()

    for state in ['S0', 'S1']:
        max_q = max(Q[state].values())

        best_actions = [
            action
            for action in actions
            if abs(Q[state][action] - max_q) < 1e-9
        ]

        policy[state] = best_actions[0]

        print(
            f"State: {state}, "
            f"Q-values: {Q[state]}, "
            f"Best action(s): {best_actions}"
        )

    # CHECK POLICY STABILITY

    if old_policy == policy:
        policy_stable = True
        print("\nPolicy is stable.")
    else:
        print("\nPolicy changed. Re-evaluating the new policy.")


# 6. Final optimal state values

print("\n4. FINAL OPTIMAL STATE VALUES")
print("-----------------------------")

for state in states:
    print(f"V*({state}) = {V[state]:.4f}")


# 7. Final action values

Q = {}

for state in ['S0', 'S1']:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]

        Q[state][action] = reward + gamma * V[next_state]

print("\n5. FINAL ACTION VALUES Q*(s, a)")
print("--------------------------------")

df_Q = pd.DataFrame.from_dict(Q, orient='index')
print(df_Q.round(4).to_string())


# 8. Display optimal policy

print("\n6. OPTIMAL POLICY")
print("-----------------")

for state in ['S0', 'S1']:
    max_q = max(Q[state].values())

    best_actions = [
        action
        for action in actions
        if abs(Q[state][action] - max_q) < 1e-9
    ]

    print(
        f"State: {state}, "
        f"Optimal Action(s): {best_actions}, "
        f"Optimal Value: {V[state]:.4f}"
    )

print("State G: Terminal")


# 9. Bellman Optimality verification

print("\n7. BELLMAN OPTIMALITY VERIFICATION")
print("----------------------------------")

for state in ['S0', 'S1']:
    max_q = max(Q[state].values())

    verified = abs(V[state] - max_q) < 1e-9

    print(
        f"State: {state}, "
        f"V*(s) = {V[state]:.4f}, "
        f"max Q*(s,a) = {max_q:.4f}, "
        f"Verified = {verified}"
    )

    assert verified

print("\nAll Bellman Optimality checks passed.")


# 10. Final result

print("\n8. RESULT")
print("---------")
print("Policy Iteration completed successfully.")
print("The optimal path is S0 -> S1 -> G.")
print("The goal state is terminal.")
print("The discount factor is 0.9.")
print("Policy Evaluation and Policy Improvement are completed.")
```
