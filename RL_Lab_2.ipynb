```python
# ============================================================
# REINFORCEMENT LEARNING - ASSIGNMENT 2
# Epsilon-Greedy Strategy for Multi-Armed Bandit
# ============================================================

import random
import pandas as pd

# ------------------------------------------------------------
# Part A - Machine Initialization
# ------------------------------------------------------------

machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

# For reproducible random numbers
random.seed(42)

# ------------------------------------------------------------
# Part B - Initialize N, R and Q
# ------------------------------------------------------------

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

total_reward = 0

results = []

print("=" * 65)
print("       EPSILON-GREEDY MULTI-ARMED BANDIT")
print("=" * 65)

print("Machines:", machines)
print("Number of Machines:", M)
print("Number of Plays:", N_play)
print("Epsilon (ε):", epsilon)

# ------------------------------------------------------------
# Part C - ε-Greedy Decision
# ------------------------------------------------------------

for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Exploration
    if r < epsilon:
        machine = random.choice(machines)
        decision = "Exploration"

    # Exploitation
    else:
        max_q = max(Q.values())

        # Find machines with highest Q-value
        best_machines = [
            m for m in machines
            if Q[m] == max_q
        ]

        machine = random.choice(best_machines)
        decision = "Exploitation"

    # --------------------------------------------------------
    # Part D - Reward Collection
    # --------------------------------------------------------

    print("\n" + "-" * 65)
    print("Play:", play)
    print("Random number (r):", round(r, 4))
    print("Decision:", decision)
    print("Selected Machine:", machine)

    # User enters reward
    reward = float(input("Enter reward obtained from Machine "
                         + machine + ": "))

    # --------------------------------------------------------
    # Part E - Q-Value Update
    # --------------------------------------------------------

    old_q = Q[machine]

    # Increase number of times machine is selected
    N[machine] += 1

    # Add reward
    R[machine] += reward

    # Incremental sample-average formula
    Q[machine] = Q[machine] + (
        (reward - Q[machine]) / N[machine]
    )

    # Add to total reward
    total_reward += reward

    print("Reward:", reward)
    print("N(" + machine + "):", N[machine])
    print("R(" + machine + "):", R[machine])

    print(
        "Q(" + machine + ") =",
        round(old_q, 4),
        "+ (",
        reward,
        "-",
        round(old_q, 4),
        ") /",
        N[machine]
    )

    print("Updated Q(" + machine + "):",
          round(Q[machine], 4))

    # Save result
    results.append([
        play,
        round(r, 4),
        decision,
        machine,
        reward,
        N[machine],
        R[machine],
        round(old_q, 4),
        round(Q[machine], 4)
    ])


# ============================================================
# Part F - FINAL N, R AND Q TABLE
# ============================================================

print("\n\n" + "=" * 65)
print("                    FINAL RESULTS")
print("=" * 65)

final_table = pd.DataFrame({
    "Machine": machines,
    "N(a) - Participation Count": [N[m] for m in machines],
    "R(a) - Total Reward": [R[m] for m in machines],
    "Q(a) - Estimated Value": [
        round(Q[m], 4) for m in machines
    ]
})

display(final_table)


# ============================================================
# PLAY-BY-PLAY RESULTS
# ============================================================

print("\n" + "=" * 65)
print("                 PLAY-BY-PLAY RESULTS")
print("=" * 65)

history_table = pd.DataFrame(
    results,
    columns=[
        "Play",
        "Random r",
        "Decision",
        "Machine",
        "Reward",
        "N(a)",
        "R(a)",
        "Old Q(a)",
        "Updated Q(a)"
    ]
)

display(history_table)


# ============================================================
# CALCULATIONS FOR TWO PLAYS
# ============================================================

print("\n" + "=" * 65)
print("                  SAMPLE CALCULATIONS")
print("=" * 65)

# Play 1
p1 = results[0]

print("\nCalculation for Play 1:")
print("Machine selected:", p1[3])
print("Reward:", p1[4])

print("Q = 0 + (", p1[4], "- 0 ) / 1")
print("Q =", round(p1[8], 4))

# Play 2
p2 = results[1]

print("\nCalculation for Play 2:")
print("Machine selected:", p2[3])
print("Reward:", p2[4])

if p2[3] == p1[3]:

    print(
        "Q =",
        p1[8],
        "+ (",
        p2[4],
        "-",
        p1[8],
        ") / 2"
    )

else:

    print(
        "Q = 0 + (",
        p2[4],
        "- 0 ) / 1"
    )

print("Q =", round(p2[8], 4))


# ============================================================
# TOTAL REWARD
# ============================================================

print("\n" + "=" * 65)
print("                    REWARD ANALYSIS")
print("=" * 65)

print("\nTotal Reward after 10 Plays =", total_reward)

print(
    "Average Reward per Play =",
    round(total_reward / N_play, 4)
)


# ============================================================
# BEST MACHINE
# ============================================================

best_machine = max(Q, key=Q.get)

print("\nBest Machine =", best_machine)
print("Highest Q-value =", round(Q[best_machine], 4))

print("\n" + "=" * 65)
print("                 PROGRAM COMPLETED")
print("=" * 65)
```
