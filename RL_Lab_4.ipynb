```python
# ============================================================
# ASSIGNMENT 4
# UPPER CONFIDENCE BOUND (UCB) - MULTI-ARMED BANDIT
# ============================================================

import math
import pandas as pd

# ------------------------------------------------------------
# PART A - MACHINE INITIALIZATION
# ------------------------------------------------------------

machines = ['A', 'B', 'C']

M = 3
N_play = 10

# N = number of times each machine is selected
N = {'A': 0, 'B': 0, 'C': 0}

# R = total reward obtained from each machine
R = {'A': 0, 'B': 0, 'C': 0}

# Q = estimated average reward
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

# UCB values
UCB = {'A': float('inf'), 'B': float('inf'), 'C': float('inf')}


# ------------------------------------------------------------
# PART B - GIVEN REWARD PATTERNS
# ------------------------------------------------------------

reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Keeps track of which reward to use next
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}


# ------------------------------------------------------------
# STORE RESULTS
# ------------------------------------------------------------

results = []


# ------------------------------------------------------------
# PART C - UCB CALCULATION AND MACHINE SELECTION
# ------------------------------------------------------------

for play in range(1, N_play + 1):

    # Calculate UCB before selecting machine
    for machine in machines:

        if N[machine] == 0:
            UCB[machine] = float('inf')

        else:
            UCB[machine] = (
                Q[machine]
                + math.sqrt(
                    (2 * math.log(play)) / N[machine]
                )
            )

    # Select machine with highest UCB
    selected_machine = max(UCB, key=UCB.get)


    # --------------------------------------------------------
    # PART D - REWARD COLLECTION
    # --------------------------------------------------------

    index = reward_index[selected_machine]

    # Cycle through the given reward pattern
    reward = reward_patterns[selected_machine][index % 4]

    reward_index[selected_machine] += 1


    # --------------------------------------------------------
    # PART E - UPDATE N, R AND Q
    # --------------------------------------------------------

    old_Q = Q[selected_machine]

    # Increase participation count
    N[selected_machine] += 1

    # Add reward
    R[selected_machine] += reward

    # Incremental sample-average formula
    Q[selected_machine] = (
        old_Q
        + (reward - old_Q) / N[selected_machine]
    )


    # --------------------------------------------------------
    # DISPLAY CALCULATION FOR FIRST TWO PLAYS
    # --------------------------------------------------------

    if play <= 2:

        print("\n========================================")
        print("CALCULATION FOR PLAY", play)
        print("========================================")

        print("Selected Machine =", selected_machine)
        print("Reward =", reward)

        print(
            "Q_new = Q_old + (Reward - Q_old) / N"
        )

        print(
            f"Q_new = {old_Q:.2f} + "
            f"({reward} - {old_Q:.2f}) / "
            f"{N[selected_machine]}"
        )

        print(
            f"Q_new = {Q[selected_machine]:.4f}"
        )


    # --------------------------------------------------------
    # CALCULATE UCB AFTER UPDATE
    # --------------------------------------------------------

    for machine in machines:

        if N[machine] == 0:
            UCB[machine] = float('inf')

        else:
            UCB[machine] = (
                Q[machine]
                + math.sqrt(
                    (2 * math.log(play)) / N[machine]
                )
            )


    # --------------------------------------------------------
    # STORE PLAY RESULT
    # --------------------------------------------------------

    results.append({
        'Play': play,
        'Machine': selected_machine,
        'Reward': reward,

        'N(A)': N['A'],
        'N(B)': N['B'],
        'N(C)': N['C'],

        'R(A)': R['A'],
        'R(B)': R['B'],
        'R(C)': R['C'],

        'UCB(A)': (
            'inf' if math.isinf(UCB['A'])
            else round(UCB['A'], 4)
        ),

        'UCB(B)': (
            'inf' if math.isinf(UCB['B'])
            else round(UCB['B'], 4)
        ),

        'UCB(C)': (
            'inf' if math.isinf(UCB['C'])
            else round(UCB['C'], 4)
        ),

        'Q(A)': round(Q['A'], 4),
        'Q(B)': round(Q['B'], 4),
        'Q(C)': round(Q['C'], 4)
    })


# ------------------------------------------------------------
# PART F - DISPLAY RESULTS
# ------------------------------------------------------------

df = pd.DataFrame(results)

print("\n========================================")
print("PLAY-BY-PLAY RESULTS")
print("========================================")

print(df.to_string(index=False))


# ------------------------------------------------------------
# FINAL N TABLE
# ------------------------------------------------------------

print("\n========================================")
print("FINAL N TABLE - PARTICIPATION COUNT")
print("========================================")

N_table = pd.DataFrame({
    'Machine': machines,
    'N (Participation Count)': [
        N['A'],
        N['B'],
        N['C']
    ]
})

print(N_table.to_string(index=False))


# ------------------------------------------------------------
# FINAL R TABLE
# ------------------------------------------------------------

print("\n========================================")
print("FINAL R TABLE - TOTAL REWARD")
print("========================================")

R_table = pd.DataFrame({
    'Machine': machines,
    'R (Total Reward)': [
        R['A'],
        R['B'],
        R['C']
    ]
})

print(R_table.to_string(index=False))


# ------------------------------------------------------------
# FINAL UCB TABLE
# ------------------------------------------------------------

print("\n========================================")
print("FINAL UCB TABLE")
print("========================================")

UCB_table = pd.DataFrame({
    'Machine': machines,
    'UCB': [
        'inf' if math.isinf(UCB['A']) else round(UCB['A'], 4),
        'inf' if math.isinf(UCB['B']) else round(UCB['B'], 4),
        'inf' if math.isinf(UCB['C']) else round(UCB['C'], 4)
    ]
})

print(UCB_table.to_string(index=False))


# ------------------------------------------------------------
# FINAL Q TABLE
# ------------------------------------------------------------

print("\n========================================")
print("FINAL Q TABLE - ESTIMATED REWARD")
print("========================================")

Q_table = pd.DataFrame({
    'Machine': machines,
    'Q (Estimated Reward)': [
        round(Q['A'], 4),
        round(Q['B'], 4),
        round(Q['C'], 4)
    ]
})

print(Q_table.to_string(index=False))


# ------------------------------------------------------------
# TOTAL REWARD RECEIVED
# ------------------------------------------------------------

total_reward = sum(R.values())

print("\n========================================")
print("TOTAL REWARD")
print("========================================")

print("Total Reward Received =", total_reward)


# ------------------------------------------------------------
# BEST MACHINE
# ------------------------------------------------------------

best_machine = max(Q, key=Q.get)

print("\n========================================")
print("BEST MACHINE")
print("========================================")

print("Machine with Highest Q-value =", best_machine)
print("Highest Q-value =", round(Q[best_machine], 4))
```
